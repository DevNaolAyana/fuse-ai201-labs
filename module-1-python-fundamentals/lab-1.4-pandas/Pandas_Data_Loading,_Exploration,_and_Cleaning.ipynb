```python
#Part 1: Import Pandas
import pandas as pd
```

Imports the Pandas library and assigns it the alias pd. Pandas provides powerful tools for data manipulation, analysis, and working with datasets in DataFrame format.

```python
#Part 2. Create DataFrame (like loading data)
data = {
    "Name": ["Sara", "Abel", "Dawit"],
    "Score": [90, 85, None]
}
df = pd.DataFrame(data)
```

Creates a pandas DataFrame from a dictionary containing student names and scores. This simulates loading a small dataset into Python.

```python
# Part 3. Data Exploration
print("First Rows:")
print(df.head())
```

head() displays the first rows.


```python
#Part 4. Dataset Information
print("\nDataset Info:")
print(df.info())
```

info() shows column names, data types, and missing values.

```python
#Part 5.Summary Of Statistics
print("\nSummary Statistics:")
print(df.describe())
```

Explores the dataset to understand its structure and contents:

describe() provides summary statistics for numeric columns.

```python
#Part 6. Detect Missing Values
print("\nMissing Values:")
print(df.isnull().sum())
```

Checks for missing (null) values in each column using isnull().sum(). This helps identify incomplete data that may need cleaning.

```python
# Part 7. Data Cleaning (fill missing values)
df["Score"] = df["Score"].fillna(df["Score"].mean())
```

Replaces missing values in the Score column with the column's mean score using fillna(). This ensures there are no empty values in the dataset.

```python
# Part 8. Show cleaned data
print("\nCleaned Data:")
print(df)
```

Displays the updated DataFrame after cleaning so you can verify that the missing values have been filled successfully.
