Example 1: Crop Yield Analysis using Matplotlib and Seaborn

```python
# Part 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Part 2. Create agriculture dataset
data = {
    "Region": ["A", "B", "C", "D", "E"],
    "Rainfall": [500, 600, 550, 700, 650],   # mm
    "Temperature": [20, 22, 19, 25, 23],     # °C
    "Yield": [2.5, 3.0, 2.8, 3.5, 3.2]       # tons per hectare
}
df = pd.DataFrame(data)
```

```python
#Part 3. Line Plot - Yield trend
plt.plot(df["Yield"])
plt.title("Crop Yield Trend by Region")
plt.xlabel("Region Index")
plt.ylabel("Yield (tons/hectare)")
plt.grid()
plt.show()
```

```python

#Part 4 Bar Chart - Compare yield across regions
plt.bar(df["Region"], df["Yield"])
plt.title("Crop Yield Comparison by Region")
plt.xlabel("Region")
plt.ylabel("Yield")
plt.show()
```

```python

# Part 5. Histogram - Yield distribution
plt.hist(df["Yield"])
plt.title("Yield Distribution")
plt.xlabel("Yield")
plt.ylabel("Frequency")
plt.show()
```

```python
# Part 6. Scatter Plot - Rainfall vs Yield
plt.scatter(df["Rainfall"], df["Yield"])
plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield")
plt.show()
```

```python
# Part 7. Seaborn Histogram
sns.histplot(df["Yield"])
plt.title("Seaborn Yield Distribution")
plt.show()
```

```python
# Part 8. Seaborn Box Plot
sns.boxplot(x=df["Yield"])
plt.title("Yield Box Plot")
plt.show()
```

```python
# Part (). Seaborn Pair Plot
sns.pairplot(df[["Rainfall", "Temperature", "Yield"]])
plt.show()
```
