# **Student Score Analysis**

```python
# 1. Import NumPy
import numpy as np
```

```python
# 2. Create data (scores of 5 students)
scores = np.array([65, 70, 75, 80, 85])

print("Student Scores:")
print(scores)
```

```python
# 3. Check number of values
print("\nNumber of scores:", scores.shape)
```

```python
# 4. Access one value (first student score)
print("\nFirst student score:", scores[3])
```

```python
# 5. Select multiple values (student 2 to student 4)
print("\nScores from student 2 to 4:", scores[1:4])
```

```python
# 6. Add 5 bonus marks to each student
new_scores = scores + 5
print("\nScores after adding 5 bonus marks:")
print(new_scores)
```

```python
# 7. Calculate statistics
print("\nAverage score:", np.mean(scores))
print("Highest score:", np.max(scores))
print("Lowest score:", np.min(scores))
print("Total score:", np.sum(scores))
```

```python
# 8. Compare each score with pass mark (50)
pass_mark = 50
difference = scores - pass_mark

print("\nDifference from pass mark:")
print(difference)
```

```python
# 9. Generate random scores (example practice data)
random_scores = np.random.randint(50, 100, size=5)

print("\nRandom practice scores:")
print(random_scores)
```
