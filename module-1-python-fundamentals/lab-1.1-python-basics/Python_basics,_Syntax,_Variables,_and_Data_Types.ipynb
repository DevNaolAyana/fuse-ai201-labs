```python
# 4. Data types
name = "Sara"       # text (string)
age = 25            # number (integer)
profit = 1200.50    # decimal (float)
is_active = True    # true/false (boolean)

print(name)
print(age)
print(profit)
print(is_active)
```

This shows different types of data in Python: text (string), numbers (int/\
float), and logical values (boolean).

```python
# 5.Type checking: find out the data type of each variable
print(type(name))
print(type(age))
print(type(profit))
print(type(is_active))
```

The type() function shows what kind of data each variable stores (e.g., string, integer, float, boolean).



```python
# 6. Simple calculation: adding value to existing variable
new_sales = sales + 10000
print("New Sales:", new_sales)
```

This adds 10,000 to the existing sales value and prints the updated result.

```python
# 7. Simple program output: displaying formatted information
print("Company:", company)
print("Total Sales:", new_sales)
```
