

# **`Example 1:Control flow example:Loan Approval System`**

```python
credit_score = 720
monthly_income = 5000
loan_amount = 20000

if credit_score >= 700 and monthly_income >= 3000:
    print("Loan Approved")
else:
    print("Loan Rejected")
```

Explanation
This is how AI models mimic decisions
We are creating rules-based intelligence
Control flow allows programs to make decisions instead of just running line by line

# **Example 3: While Loop (Condition-Based)**

```python
count = 0
while count < 3:
    print("Checking system")
    count += 1
```

Explanation:

Runs until condition becomes false
Used in:
Monitoring systems
Real-time AI pipelines

👉 “Modify loop to count how many are approved”

# **Example 7: Using Built-in Module**

```python
import math
import statistics
import datetime

# Math
print(math.sqrt(81))

# Statistics
data = [5000, 8000, 12000, 7000]
print(statistics.mean(data))

# Date
today = datetime.date.today()
print(today)
```

Explanation

math,statistics,datetime is a module

We reuse existing code instead of writing from scratch
